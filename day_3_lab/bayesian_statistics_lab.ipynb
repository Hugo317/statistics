# 0. Meta Fake News Flagger

Meta runs an algorithm that checks posts for fake news. The algorithm will make a binary decision, either to flag the content as fake news or not. 

This is a situation where deciding on what is true and not is precarious indeed. However for the purposes of this lab, we will assume the following: 

- 1% of all posts are fake news

- Metas algorithm is 80% recall

- Metas algorithm is 80% specificity


Below we define the terms for this lab:

- P(A) = Probability of fake post (base rate)

- P(B) = Probability of the algorithm returning positive (flagging for fake news)

- P(B|A) = Probability of the algorithm returning positive (flagging for fake news) given that the post truly is fake news

- P(A|B) = Probability of a fake news post given that the algorithm returns positive (flags for fake news)

# Q2 — Solving P(A|B)

## Given
- P(A) = 0.01 — base rate of fake posts
- P(B|A) = 0.80 — recall (P(flagged | truly fake))
- P(¬B|¬A) = 0.80 — specificity (P(not flagged | truly real))

## Step 1 — Derive the missing conditional
$$ P(B|\neg A) = 1 - P(\neg B|\neg A) $$

## Step 2 — Law of total probability (get P(B))
$$ P(B) = P(B|A) \cdot P(A) \;+\; P(B|\neg A) \cdot P(\neg A) $$

## Step 3 — Bayes' theorem (solve for P(A|B))
$$ P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)} $$


```python
p_a = 0.01
p_na = 0.99
p_b_a = 0.8
p_nb_na = 0.8
p_b_na  = 1 - p_nb_na
print(p_b_na)
```

```python
p_b = p_b_a * p_a + p_b_na * p_na
```

```python
print(p_b)
```

## Q2. Solve P(A|B)

P(A|B) = probability of the post actually being fake news given that the algorithm has identified it as such

```python
p_a_b = (p_b_a * p_a) / p_b
```
