# T-tests Lab

A t-test is used to compare the means of two groups. Below you'll find three different types of t-tests listed: 
- **1-sample t-test:** Used to compare the mean of a sample with a known (or made up) population mean
    - I.e.: Are people attending Nod Coding Bootcamp spending more than 40 hours per week coding?


- **Unpaired/Independent 2-sample t-test:** Used to compare the mean of two samples drawn from two separate/independent populations
    - I.e. are people who have attended Nod Coding Bootcamp spending more time coding than people who have attended an online Coding Bootcamp


- **Paired/Dependent 2-sample t-test:** Most commonly used to compare the mean of two samples drawn from the same population but where each subject/entity in the sample have been measured twice, usually before and after some treatment. (There are other variations to when a paired t-test is used but let's not get into those!)
    - I.e. are people who have attended Nod Coding Bootcamp spending more time coding after the bootcamp compared to how much time they spent coding before having attended the Bootcamp. 
    
    
The python library scipy provides access to functions with the above mentioned t-tests. Follow the link to see an overview of the statistical tests available in the scipy library - https://docs.scipy.org/doc/scipy/reference/stats.html#statistical-tests

Each t-test function in scipy returns a T-statistic and the associated p-value. 

Read more about t-tests [here](https://www.statisticshowto.com/probability-and-statistics/t-test/) and [here](https://www.investopedia.com/terms/t/t-test.asp) and/or google on your own!

```python
dof = len(df) - 1 
dof
```

## Constructing Confidence Intervals

Confidence intervals is a closely related concept to t-tests. A confidence interval gives us an estimate of an interval containing the true mean of the population. I.e. for a 95% confidence interval, we are 95% sure that the true mean of the population is within the esitmated confidence interval. 

To read more about confidence intervals, click [here](https://en.wikipedia.org/wiki/Confidence_interval) and/or google on your own

## 1. Construct a 95% CI for the mean hourly wage of all hourly workers

We will construct a 95% confidence interval for the mean hourly wage of all hourly workers. 

A function for computing the confidence interval can be found in the SciPy library with the name `t.interval`. You can find the documentation for this function [here](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.t.html).

To compute the 95% confidence interval for the hourly wage:
- use 0.95 for the confidence level/alpha level 
- (number of rows - 1) for degrees of freedom
- the mean of the sample for the location parameter 
- the standard error for the scale. The standard error can be computed using [this](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.sem.html) function in the SciPy library.

```python
data = df["Hourly Rate"].dropna()

interval = stats.t.interval(
    0.95,
    df=len(data) - 1,
    loc=data.mean(),
    scale=stats.sem(data)
)
interval
```

## 2. Construct a 95% CI for the mean annual salary for all salaried employees in the Police Department

```python
data = df.loc[df["Department"] == "POLICE"]["Annual Salary"].dropna()

interval = stats.t.interval(
    0.95,
    df=len(data) - 1,
    loc=data.mean(),
    scale=stats.sem(data)
)
interval
```
