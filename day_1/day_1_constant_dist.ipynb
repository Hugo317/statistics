## Central limit theorem 

```python
die = [1,2,3,4,5,6]
```

```python
# mean 1 sample with 100 throws 
sample_lst = []
for i in range(100):
    sample_lst.append(np.random.choice(die))
```

```python
np.mean(sample_lst)
```

```python
no_of_throws = 100

sample_mean_lst = []

for i in range (100000):
    tempt_lst = []
    for j in range(no_of_throws):
        throw = np.random.choice(die)
        tempt_lst.append(throw)

        #calculate and append mean of 1 sample with n no. of throws
    sample_mean = np.array(tempt_lst).mean()

    sample_mean_lst.append(sample_mean)
```

```python
sns.kdeplot(sample_mean_lst)
```

## Standard error
<br>
 Standard error is the standard deviation of its sampling distribution or an estimate of that standard deviation.
<br>
$$Standard \ Error = \frac{\sigma}{\sqrt{n}} $$
<br>
sigma = Sample Mean <br>
n=Sample Size


```python
sigma = np.array(die).std()
```

```python
sqrt_n = np.sqrt(no_of_throws)
```

```python
se_100_throws = sigma / sqrt_n
```

```python
se_100_throws
```

```python
np.std(sample_mean_lst)
```

<br>
The standard error (SE) measures the accuracy with which a sample represents a population. Specifically, it's the standard deviation of the sampling distribution of a statistic, most commonly the mean. It tells us how much the sample mean (or other statistic) is expected to vary from the true population mean.
<br>

$$Standard \ Error = \frac{\sigma}{\sqrt{n}} $$

<br>
$\sigma = Sample \ Standard Deviation$ <br>
$n=Sample \ Size$

```python
sample_mean = np.random.normal(iq_dist)
sample_std = np.std(iq_dist)
sample_size = len(iq_dist)
```

```python
satandard_error = sample_std / np.sqrt(sample_size)
print(f"Sample Mean: {sample_mean}")
print(f"Sample Standard: {sample_std}")
print(f"Sample Size: {sample_size}")
print(f"Standard Error: {satandard_error}")

```
