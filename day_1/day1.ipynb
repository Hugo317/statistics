# Discrete distributions

1. Uniform:
    - manually,
    - `np.random`
    - visualization `sns.histplot()`
2. Binomial:
    - manually
    - `np.random`
    - visualization `sns.histplot()`
3. Poisson:
    - `np.random`
    - visualization `sns.histplot()`

```python
import numpy as np
import random
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
die = [i for i in range(1,7)]
```

```python
random.choice(die)
```

```python
outcome = []
for i in range(10):
    outcome.append(random.choice(die))
```

```python
outcome
```

```python
sns.histplot(outcome,discrete= True)
```

```python
outcome2 = []
for i in range(100000):
    outcome2.append(random.choice(die))
```

```python
sns.histplot(outcome2,discrete= True)
```

```python
coin = ["heads","tails"]
```

```python
outcome_coin = []
for i in range(10):
    outcome_coin.append(random.choice(coin))
```

```python
outcome_coin.count("tails") / len(outcome_coin)
```

```python
np.random.binomial(n = 10,p= 0.5 )
```

```python
test_outcome = np.random.binomial(n = 10,p = 0.5, size = 10000)
sns.histplot(test_outcome, discrete=True)
```

```python
outcome = np.random.binomial(n = 10, p = 0.5, size = 10000)
plot = sns.histplot(outcome, discrete = True, stat = "density")

for p in plot.patches:

    if p.get_height() == 0:
        pass
    else:
        plot.annotate(format(p.get_height(), '.2f'),
        (p.get_x() + p.get_width() / 2., p.get_height()),
        ha = 'center', va = 'center',
        xytext = (0, 9),
        textcoords = 'offset points')

plt.margins(y = 0.1)
plt.show()
```
